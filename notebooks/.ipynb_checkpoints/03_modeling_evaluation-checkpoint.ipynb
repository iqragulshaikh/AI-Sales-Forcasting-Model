# Import and setup

```python
import os
import sys
from pathlib import Path


notebook_dir = Path.cwd()
root_dir = notebook_dir.parent

if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
from src_.preprocessing import prepare_sales_data
from src_.features import create_features
from src_.baseline import make_baseline_forecast

print("SUCCESS: Modules successfully loaded")
```

```python
excel_path = root_dir / "data_" / "sample_sales_dataset (2).xlsx"
df_raw = pd.read_excel(excel_path)

df_clean = prepare_sales_data(df_raw) if 'prepare_sales_data' in globals() else df_raw.copy()
df_features = create_features(df_clean)

print("Dataset ke Columns yeh hain:")
print(df_features.columns.tolist())
sort_cols = []
if 'branch_id' in df_features.columns:
    sort_cols.append('branch_id')
elif 'branch' in df_features.columns:
    sort_cols.append('branch')
elif 'store_id' in df_features.columns:
    sort_cols.append('store_id')

if 'date' in df_features.columns:
    df_features['date'] = pd.to_datetime(df_features['date'])
    sort_cols.append('date')

if sort_cols:
    df_features = df_features.sort_values(by=sort_cols).reset_index(drop=True)

print("\nDataset Shape:", df_features.shape)
df_features.head()
```

```python
target = 'sales'
branch_col = 'branch_id' if 'branch_id' in df_features.columns else ('branch' if 'branch' in df_features.columns else 'store_id')

ignore_cols = ['date', branch_col, 'sales']
feature_cols = [col for col in df_features.columns if col not in ignore_cols and pd.api.types.is_numeric_dtype(df_features[col])]

max_date = df_features['date'].max()
split_date = max_date - pd.Timedelta(days=7)

train_df = df_features[df_features['date'] <= split_date].copy()
test_df = df_features[df_features['date'] > split_date].copy()

train_df = train_df.dropna(subset=feature_cols)
test_df = test_df.dropna(subset=feature_cols)

X_train, y_train = train_df[feature_cols], train_df[target]
X_test, y_test = test_df[feature_cols], test_df[target]

print(f"Train period: {train_df['date'].min().date()} to {train_df['date'].max().date()}")
print(f"Test period:  {test_df['date'].min().date()} to {test_df['date'].max().date()}")
print("Features count:", len(feature_cols))
```

 # Main Modules Run & Baseline Comparision

```python
#Baseline Model Execution (Fixed Function Call)
try:
    if 'branch_id' in train_df.columns:
        b_id = train_df['branch_id'].iloc[0]
        baseline_df = make_baseline_forecast(train_df, branch_id=b_id)
        baseline_pred = baseline_df['predicted_sales'].values[:len(y_test)]
    else:
        raise KeyError("'branch_id' column not found in raw format")
except Exception:

    if 'sales_lag_7' in test_df.columns and not test_df['sales_lag_7'].isna().all():
        baseline_pred = test_df['sales_lag_7'].fillna(train_df['sales'].mean())
    else:
        baseline_pred = np.full(len(y_test), train_df['sales'].mean())

base_mae = mean_absolute_error(y_test, baseline_pred)
base_rmse = np.sqrt(mean_squared_error(y_test, baseline_pred))

# Main Model
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

test_df['predicted_sales'] = model.predict(X_test)
rf_mae = mean_absolute_error(y_test, test_df['predicted_sales'])
rf_rmse = np.sqrt(mean_squared_error(y_test, test_df['predicted_sales']))

# Model Comparison Output

print("       BASELINE VS MAIN MODEL EVALUATION  ")

print(f"Baseline Model MAE : {base_mae:.2f}")
print(f"Baseline Model RMSE: {base_rmse:.2f}"
     )
print("------------------------------------------")

print(f"Random Forest MAE : {rf_mae:.2f}")
print(f"Random Forest RMSE: {rf_rmse:.2f}")
```

```python
# 1. Reconstruct original 'branch_id' from one-hot encoded columns if missing
if 'branch_id' not in test_df.columns:
    if 'branch_B001' in test_df.columns:
        # Extract branch name from one-hot encoded columns (e.g., branch_B001 -> B001)
        branch_cols = [c for c in test_df.columns if c.startswith('branch_')]
        if branch_cols:
            test_df['branch_id'] = test_df[branch_cols].idxmax(axis=1).str.replace('branch_', '')
        else:
            test_df['branch_id'] = 'B001'
    elif 'branch' in test_df.columns:
        test_df['branch_id'] = test_df['branch']
    elif 'store_id' in test_df.columns:
        test_df['branch_id'] = test_df['store_id']
    else:
        test_df['branch_id'] = 'B001'

# 2. Extract standardized forecast table
forecast_output = test_df[['date', 'branch_id', 'predicted_sales']].copy()

# 3. Format Date and Sales numbers
if pd.api.types.is_datetime64_any_dtype(forecast_output['date']):
    forecast_output['date'] = forecast_output['date'].dt.strftime('%Y-%m-%d')

forecast_output['predicted_sales'] = forecast_output['predicted_sales'].round(2)

# Ensure correct column ordering according to team contract
forecast_output = forecast_output[['date', 'branch_id', 'predicted_sales']]
```
