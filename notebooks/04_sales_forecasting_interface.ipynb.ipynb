### Mini-Plan:

2. **Interactive Query Function:** Build a function that takes `branch_id` and `date` to return sales predictions.
3. **Input Validation & Error Handling:** Handle edge cases like invalid branch IDs, incorrect date formats, and out-of-range dates gracefully.
4. **Testing & Logging:** Run 10 comprehensive test cases (valid and invalid inputs) and document the results.

```python
import pandas as pd
import numpy as np
from datetime import datetime

# Load the generated handoff/forecast dataset
DATA_PATH = "AI-Sales-Forcasting-Model/data/forecast_handoff_example.csv"

try:
    df_sales = pd.read_csv(DATA_PATH)
    df_sales['date'] = pd.to_datetime(df_sales['date'])
    print("✓ Dataset loaded successfully!")
    print(f"Total Records: {len(df_sales)}")
    display(df_sales.head())
except Exception as e:
    print(f"Error loading dataset: {e}")
```

```python
def predict_branch_sales(branch_id, target_date, data=df_sales):
    """
    Handles user queries for sales predictions with robust input validation.
    """
    # 1. Validate Branch ID type and existence
    if not isinstance(branch_id, (int, str)) or str(branch_id).strip() == "":
        return {"success": False, "error": "Invalid Branch ID: Branch ID cannot be empty."}
    
    # 2. Validate Date Format
    try:
        parsed_date = pd.to_datetime(target_date)
    except Exception:
        return {"success": False, "error": f"Invalid Date Format: '{target_date}'. Use YYYY-MM-DD."}
    
    # 3. Filter DataFrame
    branch_matches = data[data['branch_id'].astype(str) == str(branch_id)]
    
    if branch_matches.empty:
        valid_branches = data['branch_id'].unique().tolist()
        return {
            "success": False, 
            "error": f"Branch '{branch_id}' not found. Available branches: {valid_branches}"
        }
    
    date_match = branch_matches[branch_matches['date'] == parsed_date]
    
    if date_match.empty:
        min_date = branch_matches['date'].min().strftime('%Y-%m-%d')
        max_date = branch_matches['date'].max().strftime('%Y-%m-%d')
        return {
            "success": False, 
            "error": f"Date '{target_date}' out of range. Available range for Branch {branch_id}: {min_date} to {max_date}"
        }
    
    # 4. Return Successful Prediction
    predicted_val = date_match['predicted_sales'].values[0]
    return {
        "success": True,
        "branch_id": branch_id,
        "date": parsed_date.strftime('%Y-%m-%d'),
        "predicted_sales": round(float(predicted_val), 2)
    }
```

```python
# Quick Test
print(predict_branch_sales(branch_id="B001", target_date="2026-10-01"))
print(predict_branch_sales(branch_id="INVALID_BRANCH", target_date="2026-10-01"))
```
